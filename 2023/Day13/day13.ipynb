```python
import os,sys
import numpy as np
import functools, collections
```

```python
def check_lines(line1, line2):
    if functools.reduce(lambda x, y : x and y, map(lambda p, q: p == q,line1,line2), True):
        return True
    return False

def check_array(inArray):
    for idx in range(len(inArray)-1):
        if check_lines(inArray[idx],inArray[idx+1]):
            i, state = 1, True
            while (idx-i) >= 0 and (idx+1+i) < len(inArray):
                if not check_lines(inArray[idx-i],inArray[idx+1+i]):
                    state = False
                    break
                i+=1
            if state:
                return idx+1
    return 0

def readFile(inFile):
    sketch_dict, idx = {}, 0
    with open(inFile, 'r') as f:
        sketch_dict[idx] = []
        for l in f:
            if l == '\n':
                idx += 1
                sketch_dict[idx] = list()
            else:     
                sketch_dict[idx].append([char for char in l.strip()]) 

    for i in range(idx):
        sketch_dict[i] = np.asarray(sketch_dict[i]) 
    return sketch_dict

def main(inFile):
    source_dict = readFile(inFile)
    horz, vert = 0, 0
    for key in source_dict.keys():
        horz += check_array(source_dict[key])
        vert += check_array(np.transpose(source_dict[key]))
    return horz*100 + vert
```
