```python
import os, re
import numpy as np
```

```python
def read_file(inputFile:str) -> (list, dict, dict):
    l_dict, r_dict, pos_dict = {}, {}, {}
    with open(inputFile, 'r') as inFile:
        instructions = [char for char in inFile.readline().strip()]
        inFile.readline()
        line = inFile.readline()
        while True:
            if not line:
                break
            result = re.search(r"(\w\w(\w)) = \(([\w]{3}), ([\w]{3})\)", line)
            if result[2] == 'A':
                pos_dict[result[1]] = result[1]
            l_dict[result[1]] = result[3]
            r_dict[result[1]] = result[4]
            ## print(line, result[1], result[2], result[3])
            line = inFile.readline()
        return instructions, l_dict, r_dict, pos_dict
```

```python
inst, ldict, rdict, pos_dict = read_file('../Input/day8.txt')
```

```python
def move_instr(pos:str, instructions:list, ind) -> str:
    if inst[ind] == 'R':
        return rdict[pos] 
    if inst[ind] == 'L':
        return ldict[pos]
```

```python
def check_val(pos:str) -> bool:
    if len(pos) < 2:
        return False
    if pos[2] == 'Z':
        return True
    return False
```

```python
pos_dict = {key:key for key in pos_dict.keys()}
print(pos_dict)
```

```python
print(pos_dict)
print(False in map(check_val, pos_dict.values()))
print(True in map(check_val, pos_dict.values()))

## Look into LCM and Repeats on different terminations
```

```python
## for key, val in pos_dict, find the termini and the # steps
termini_dict = {key:[] for key in pos_dict.keys()}
for key in pos_dict.keys():
    ind, node_list, counter = 0, [], 0
    while True:
        new_pos = move_instr(pos_dict[key], inst, ind)
        counter += 1
        node_list.append(new_pos)
        pos_dict[key] = new_pos
        if check_val(new_pos):
            termini_dict[key].append([new_pos,counter])
            break
        ind += 1
        ind %= len(inst)
print(pos_dict)
print(termini_dict)
```
