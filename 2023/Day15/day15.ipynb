```python
def hash_map(inStr):
    start_val = 0
    for char in inStr:
        start_val += ord(char)
        start_val *= 17
        start_val %= 256
    return start_val

## sub is letters
def list_remove(sub,super):
    for lens in super:
        if sub == lens[0]:
            super.remove(lens)
    return super

def list_contain(sub, super):
    for lens in super:
        if sub == lens[0]:
            return True
    return False

def list_ind(sub, super):
    for idx in range(len(super)):
        if sub == super[idx][0]:
            return idx
        
def focus(inList):
    return sum([(idx+1)*int(val[1]) for idx, val in enumerate(inList)])
        

```

```python
##pt 1, hash map works
total_sum = 0
for inStr in hashList:
    total_sum += hash_map(inStr)
print(total_sum)

sum(map(hash_map,hashList))
```

```python
boxes = collections.defaultdict()
for i in range(256):
    boxes[i] = []

for inStr in hashList:
    if '-' in inStr:
        pts = inStr.split('-')
        box = hash_map(pts[0])
        if boxes[box] == []:
            continue
        list_remove(pts[0],boxes[box])
    if '=' in inStr:
        pts = inStr.split('=')
        box, lens = hash_map(pts[0]), pts[1]
        if list_contain(pts[0], boxes[box]):
            idx = list_ind(pts[0], boxes[box])
            boxes[box][idx] = [pts[0], lens]
        else:
            boxes[box].append([pts[0], lens])
    
```

```python
for i in range(10):
    print(boxes[i])
```

```python
total_focus = 0
for i in range(256):
    total_focus += (i+1)*focus(boxes[i])
print(total_focus)
```
