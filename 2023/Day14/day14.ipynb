```python
import os, sys
import numpy as np 
import time
```

```python
def readFile(inFile):
    outarr = []
    with open(inFile, 'r') as f:
        for line in f.readlines():
            outarr.append([char for char in line.strip()])
    return np.asarray(outarr)

## convert to array logic
## inArr[i][j]
## provide a direction (N, W, S, E)
## N/S inverse on columns, W/E inverse on rows
def adjust_row(inrow):
    stepper = 0
    for i in range(len(inrow)):
        if inrow[i] == '.':
            stepper += 1
        if inrow[i] == '#':
            stepper = 0
        if inrow[i] == 'O':
            inrow[i-stepper] = 'O'
            if i - stepper != i:
                inrow[i] = '.'
    return inrow

def adjust_table(inArr, direction):
    width = inArr.shape[0] ## 100
    height = inArr.shape[1] ## 100
    if direction == 'W':
        for i in range(width):
            stepper = 0
            for j in range(height):
                if inArr[i][j] == '.':
                    stepper += 1
                if inArr[i][j] == '#':
                    stepper = 0
                if inArr[i][j] == 'O':
                    inArr[i][j-stepper] = 'O'
                    if j - stepper != j:
                        inArr[i][j] = '.'
    if direction == 'E':
        for i in range(width):
            stepper = 0
            for jx in range(height):
                j = height - 1 - jx
                if inArr[i][j] == '.':
                    stepper += 1
                if inArr[i][j] == '#':
                    stepper = 0
                if inArr[i][j] == 'O':
                    inArr[i][j+stepper] = 'O'
                    if j + stepper != j:
                        inArr[i][j] = '.'
    if direction == 'N':
        for j in range(height):
            stepper = 0
            for i in range(width):
                if inArr[i][j] == '.':
                    stepper += 1
                if inArr[i][j] == '#':
                    stepper = 0
                if inArr[i][j] == 'O':
                    inArr[i-stepper][j] = 'O'
                    if i - stepper != i:
                        inArr[i][j] = '.'
    if direction == 'S':
        for j in range(height):
            stepper = 0
            for ix in range(width):
                i = width-1 - ix
                if inArr[i][j] == '.':
                    stepper += 1
                if inArr[i][j] == '#':
                    stepper = 0
                if inArr[i][j] == 'O':
                    inArr[i+stepper][j] = 'O'
                    if i + stepper != i:
                        inArr[i][j] = '.'
    return inArr

def total_weight(inArr, direction):
    total_weight = 0
    for ro in range(len(inArr)):
        if direction == 'N':
            correction = len(inArr) - ro
        elif direction == 'S':
            correction = ro
        total_weight += (correction)*sum([1 for i in inArr[ro] if i == "O" ])
    return total_weight
```

```python
inArr = readFile('../Input/day14.txt')
print(inArr)
start = time.time()
iter_list, idx = [], 0
for i in range(10000):
    inArr = adjust_table(adjust_table(adjust_table(adjust_table(inArr, "N"),"W"),"S"),"E")
    iter_list.append(total_weight(inArr,"N"))
end = time.time()
print(f"total time is: {end - start}")

```
