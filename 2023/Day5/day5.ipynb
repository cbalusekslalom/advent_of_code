```python
def read_in_file(inputfile):
    seed_dict = {}
    out_dict = {}
    seed_list = []
    with open(inputfile, 'r') as inFile:
        line = inFile.readline()
        while line:
            print(line)
            if 'seeds' in line:
                key, lst = line.split(': ')
                seed_dict = {val:int(val) for val in lst.strip().split(' ')}
                seed_list = [int(val) for val in lst.strip().split(' ')]
            elif 'map' in line:
                key, out_dict[key] = line.split(' ')[0], []
                line = inFile.readline().strip()
                while line != '\n':
                    out_dict[key].append([int(val) for val in line.strip().split(' ')])
                    line = inFile.readline()
                    if line == '':
                        break
            line = inFile.readline()
    return seed_dict, out_dict, seed_list 
            
```
