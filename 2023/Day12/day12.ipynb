```python
import os, sys
```

```python
def readFile(inFile):
    line_dict, idx = {}, 0
    with open(inFile, 'r') as f:
        for l in f:
            line = l.split(' ')
            line_dict[idx] = {'sequence': [char for char in line[0]], 'order': [int(num) for num in line[1].strip().split(',')]}
            idx += 1
    return line_dict
```

```python
sample_dict = readFile('../Input/day12_sample.txt')
```
